Oppgave A

For å finne de ti største landene bruker jeg bibliotekt collections

```python
import json
from collections import Counter
import matplotlib.pyplot as plt

filnavn = "Global YouTube Statistics.json"

with open(filnavn, encoding="utf-8") as fil:
    data = json.load(fil)

land = []

for item in data:
    land.append(item["Country"])

counter = Counter(land)

most_common = counter.most_common()    

x_verdier = ["US", "India", "nan", "Brazil", "UK", "Mexico", "Indo", "Spain", "Thai", "S Korea"]
y_verdier = []

for i in range(10):
    y_verdier.append(most_common[i][1])

print(x_verdier)
print(y_verdier)

plt.bar(x_verdier, y_verdier)
plt.show()

        
```

En annen versjon

```python
import json
import matplotlib.pyplot as plt

filnavn = "Global YouTube Statistics.json"

with open(filnavn, encoding="utf-8") as fil:
    data = json.load(fil)

land = {}

for item in data:
    i = 0
    for landa in data:
        key = item["Country"]
        if key == landa["Country"]:
            i += 1
    land[item["Country"]] = i    

print(land)

landene = list(land)

print(landene)

x2_verdier = []
y2_verdier = []

for i in range(10):
    max_count_country = max(land, key=land.get)
    x2_verdier.append(max_count_country)
    y2_verdier.append(land[max_count_country])
    del land[max_count_country]

print(landene)

plt.bar(x2_verdier, y2_verdier)
plt.show()

```

Oppgave B

Antall abonnenter per land

```python
gjennomsnitt_aper_land = []

for i in range(10):
    sum_abonnenter = 0
    sum_videovisninger = 0
    for item in data:
        if item["Country"] == most_common[i][0]:  
            sum_abonnenter += item["subscribers"] 
            sum_videovisninger += item["video views"]
    gjennomsnitt = sum_abonnenter/most_common[i][1]
    gjennomsnitt_aper_land.append(round(gjennomsnitt))
    
print(gjennomsnitt_aper_land)


plt.bar(x_verdier, gjennomsnitt_aper_land)
plt.title("Gjennomsnittlig abonnenter per land")
plt.ylabel("Antall abonnenter")
plt.xlabel("Land")
plt.show()

```
